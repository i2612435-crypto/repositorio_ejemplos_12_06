```python
# IDL2 - Ejercicio 5: Interés bancario
# Estructura secuencial

print("=== EJERCICIO 5: INTERÉS BANCARIO ===")
capital = 10000
tasa_mensual = 0.02  # 2%
ganancia = capital * tasa_mensual * 12

print(f"Capital inicial: S/{capital}")
print(f"Tasa mensual: {tasa_mensual * 100}%")
print(f"Ganancia en 1 año: S/{ganancia:.2f}")
```
