```python
# IDL2 - Ejercicio 13: Verificar tributación
# Estructura condicional

print("=== EJERCICIO 13: TRIBUTACIÓN ===")
edad = 20
ingresos = 3500

if edad > 18 and ingresos >= 3000:
    tributo = ingresos * 0.05
    print(f"Edad: {edad} años")
    print(f"Ingresos: S/{ingresos}")
    print(f"Debe tributar: S/{tributo:.2f} (5%)")
else:
    print("No debe tributar")
```
