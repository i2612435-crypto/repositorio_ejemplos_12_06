```python
# IDL2 - Ejercicio 18: Ordenar 3 números
# Estructura condicional

print("=== EJERCICIO 18: ORDENAR NÚMEROS ===")
a, b, c = 15, 3, 8
orden = "ascendente"  # Cambiar a "descendente" si se desea

numeros = [a, b, c]
if orden == "ascendente":
    numeros.sort()
else:
    numeros.sort(reverse=True)

print(f"Números originales: {a}, {b}, {c}")
print(f"Orden {orden}: {numeros}")
```
