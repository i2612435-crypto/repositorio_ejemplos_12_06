```python
# IDL2 - Ejercicio 14: Colegiatura con descuento
# Estructura condicional

print("=== EJERCICIO 14: COLEGIATURA ===")
materias = 6
promedio = 19
costo_materia = 100

total = materias * costo_materia
if promedio >= 18:
    descuento = (total / 2) * 0.5
    total -= descuento
    print(f"Promedio: {promedio} → Aplica descuento del 50% sobre mitad de cursos")
else:
    print(f"Promedio: {promedio} → Sin descuento")

print(f"Total a pagar: S/{total}")
```
