```python
# Ejercicio 23: Porcentajes respecto a un número X
numeros = [3, 7, 9, 2, 10, 4, 5, 8]
X = 5
total = len(numeros)
menores = mayores = iguales = entre = ninguna = 0

for n in numeros:
    if n < X: menores += 1
    if n > X: mayores += 1
    if n == X: iguales += 1
    if 1 <= n <= X: entre += 1
    if not (n < X or n > X or n == X or (1 <= n <= X)): ninguna += 1

print(f"Menores que {X}: {(menores/total)*100:.1f}%")
print(f"Mayores que {X}: {(mayores/total)*100:.1f}%")
print(f"Iguales a {X}: {(iguales/total)*100:.1f}%")
print(f"Entre 1 y {X}: {(entre/total)*100:.1f}%")
print(f"Ninguna condición: {(ninguna/total)*100:.1f}%")
```
