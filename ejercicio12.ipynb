```python
# IDL2 - Ejercicio 12: Mayor y menor de 3 números
# Estructura condicional

print("=== EJERCICIO 12: MAYOR Y MENOR ===")
a, b, c = 45, 12, 78

mayor = max(a, b, c)
menor = min(a, b, c)

print(f"Números: {a}, {b}, {c}")
print(f"Número mayor: {mayor}")
print(f"Número menor: {menor}")
```
