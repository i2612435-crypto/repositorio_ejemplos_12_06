```python
# IDL2 - Ejercicio 17: Llamada telefónica con impuesto
# Estructura condicional

print("=== EJERCICIO 17: LLAMADA TELEFÓNICA ===")
minutos = 12
es_domingo = True
es_turno_nocturno = False

# Costo base
if minutos <= 5:
    costo = minutos * 1.0
elif minutos <= 8:
    costo = 5*1.0 + (minutos-5)*0.8
elif minutos <= 10:
    costo = 5*1.0 + 3*0.8 + (minutos-8)*0.7
else:
    costo = 5*1.0 + 3*0.8 + 2*0.7 + (minutos-10)*0.5

# Impuesto
if es_domingo and es_turno_nocturno:
    impuesto = 0.05
elif es_domingo:
    impuesto = 0.03
else:
    impuesto = 0

total = costo * (1 + impuesto)
print(f"Duración: {minutos} minutos")
print(f"Costo base: S/{costo:.2f}")
print(f"Impuesto: {impuesto*100}%")
print(f"Total: S/{total:.2f}")
```
