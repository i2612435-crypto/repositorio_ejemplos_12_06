```python
# IDL2 - Ejercicio 9: Compra con descuento y cambio
# Estructura secuencial

print("=== EJERCICIO 9: COMPRA CON DESCUENTO ===")
monto_base = 250
descuento = 10  # 10%
pago_efectivo = 300

monto_final = monto_base * (1 - descuento/100)
cambio = pago_efectivo - monto_final

print(f"Monto base: S/{monto_base}")
print(f"Descuento: {descuento}%")
print(f"Monto final con descuento: S/{monto_final:.2f}")
print(f"Pago con efectivo: S/{pago_efectivo}")
print(f"Cambio: S/{cambio:.2f}")
```
