```python
# IDL2 - Ejercicio 10: Cajero automático
# Estructura secuencial

print("=== EJERCICIO 10: CAJERO AUTOMÁTICO ===")
monto = 387
denominaciones = [200, 100, 50, 20, 10, 5, 1]

print(f"Monto a retirar: S/{monto}")
print("Desglose:")
monto_temp = monto
for d in denominaciones:
    cantidad = monto_temp // d
    if cantidad > 0:
        print(f"  {d}: {cantidad}")
    monto_temp %= d
```
