```python
# Ejercicio 22: Contar pares, impares, positivos, neutros, negativos
numeros = [5, -3, 0, 8, -1, 4, 0, 7, -2, 6]
pares = impares = positivos = neutros = negativos = 0

for n in numeros:
    if n % 2 == 0:
        pares += 1
    else:
        impares += 1

    if n > 0:
        positivos += 1
    elif n == 0:
        neutros += 1
    else:
        negativos += 1

print(f"Pares: {pares}, Impares: {impares}")
print(f"Positivos: {positivos}, Neutros: {neutros}, Negativos: {negativos}")
```
