```python
# IDL2 - Ejercicio 4: Comisión de vendedor
# Estructura secuencial

print("=== EJERCICIO 4: COMISIÓN DE VENDEDOR ===")
sueldo_base = 1200
ventas = [800, 600, 1000]
comision = sum(ventas) * 0.25
total = sueldo_base + comision

print(f"Sueldo base: S/{sueldo_base}")
print(f"Ventas: {ventas}")
print(f"Comisión (25%): S/{comision:.2f}")
print(f"Total a recibir: S/{total:.2f}")
```
