```python
# Ejercicio 27: Factorial de dos números
def factorial(n):
    resultado = 1
    for i in range(1, n+1):
        resultado *= i
    return resultado

a, b = 7, 10
print(f"Factorial de {a}: {factorial(a)}")
print(f"Factorial de {b}: {factorial(b)}")
print(f"Son iguales hasta el múltiplo {min(a, b)}")
```
