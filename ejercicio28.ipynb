```python
# Ejercicio 28: Estadísticas de asistentes a una fiesta
asistentes = [("M", 25), ("F", 30), ("M", 22), ("F", 28), ("M", 35), ("F", 27)]

total = len(asistentes)
hombres = sum(1 for s, _ in asistentes if s == "M")
mujeres = total - hombres

suma_h = sum(e for s, e in asistentes if s == "M")
suma_m = sum(e for s, e in asistentes if s == "F")

prom_h = suma_h / hombres if hombres > 0 else 0
prom_m = suma_m / mujeres if mujeres > 0 else 0

edades = [e for _, e in asistentes]

print(f"Total asistentes: {total}")
print(f"Hombres: {hombres}, Mujeres: {mujeres}")
print(f"Promedio edad hombres: {prom_h:.1f}")
print(f"Promedio edad mujeres: {prom_m:.1f}")
print(f"Menor edad: {min(edades)}, Mayor edad: {max(edades)}")
```
