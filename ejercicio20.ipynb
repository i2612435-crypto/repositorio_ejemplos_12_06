```python
# IDL2 - Ejercicio 20: Llamada internacional por zona
# Estructura condicional múltiple

print("=== EJERCICIO 20: LLAMADA INTERNACIONAL ===")
clave = 20
minutos = 10

if clave == 10:
    zona = "América del norte"
    costo_x_min = 2.2
elif clave == 12:
    zona = "América del centro"
    costo_x_min = 2.5
elif clave == 20:
    zona = "América del sur"
    costo_x_min = 1.2
elif clave == 22:
    zona = "Asia"
    costo_x_min = 3.5
elif clave == 30:
    zona = "Europa"
    costo_x_min = 3.0
elif clave == 32:
    zona = "África"
    costo_x_min = 3.2
else:
    zona = "Desconocida"
    costo_x_min = 0

total = costo_x_min * minutos
print(f"Clave: {clave}")
print(f"Zona: {zona}")
print(f"Costo por minuto: S/{costo_x_min}")
print(f"Duración: {minutos} minutos")
print(f"Total: S/{total}")
```
