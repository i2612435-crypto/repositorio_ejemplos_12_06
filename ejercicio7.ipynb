```python
# IDL2 - Ejercicio 7: Cobro de estacionamiento
# Estructura secuencial

print("=== EJERCICIO 7: ESTACIONAMIENTO ===")
horas = 3
minutos = 20
horas_totales = horas + (1 if minutos > 0 else 0)
cobro = horas_totales * 5

print(f"Tiempo estacionado: {horas} horas y {minutos} minutos")
print(f"Horas cobradas (fracción completa): {horas_totales}")
print(f"Cobro total: S/{cobro}")
```
