```python
# IDL2 - Ejercicio 19: Descuento por cantidad de camisas
# Estructura condicional

print("=== EJERCICIO 19: DESCUENTO CAMISAS ===")
n = 7
precio = 50

total_sin_desc = n * precio

if 1 <= n <= 4:
    descuento = 0.125
elif 5 <= n <= 8:
    descuento = 0.20
else:
    descuento = 0.315

monto_descuento = total_sin_desc * descuento
total_con_desc = total_sin_desc - monto_descuento

print(f"Cantidad de camisas: {n}")
print(f"Precio unitario: S/{precio}")
print(f"Total sin descuento: S/{total_sin_desc}")
print(f"Descuento: {descuento*100}% → S/{monto_descuento:.2f}")
print(f"Total con descuento: S/{total_con_desc:.2f}")
```
