```python
# Ejercicio 25: Ahorro acumulado durante 12 meses
depositos = [100, 150, 200, 130, 180, 210, 160, 140, 170, 190, 220, 240]
acumulado = 0

for mes, monto in enumerate(depositos, 1):
    acumulado += monto
    print(f"Mes {mes}: lleva ahorrado S/{acumulado}")

print(f"Ahorro final del año: S/{acumulado}")
```
