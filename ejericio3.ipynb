```python
# IDL2 - Ejercicio 3: Porcentaje de inversión
# Estructura secuencial

print("=== EJERCICIO 3: PORCENTAJE DE INVERSIÓN ===")
inv1, inv2, inv3 = 1500, 2500, 4000
total = inv1 + inv2 + inv3

p1 = (inv1 / total) * 100
p2 = (inv2 / total) * 100
p3 = (inv3 / total) * 100

print(f"Inversión Persona 1: S/{inv1} -> {p1:.1f}%")
print(f"Inversión Persona 2: S/{inv2} -> {p2:.1f}%")
print(f"Inversión Persona 3: S/{inv3} -> {p3:.1f}%")
print(f"Total invertido: S/{total}")
```
