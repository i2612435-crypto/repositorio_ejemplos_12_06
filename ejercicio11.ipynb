```python
# IDL2 - Ejercicio 11: Precio de llantas según cantidad
# Estructura condicional

print("=== EJERCICIO 11: LLANTERAS ===")
cantidad = 4

if cantidad < 5:
    precio = 150
else:
    precio = 120

total = cantidad * precio
print(f"Cantidad de llantas: {cantidad}")
print(f"Precio por llanta: S/{precio}")
print(f"Total a pagar: S/{total}")
```
