```python
# Ejercicio 30: Estadísticas completas de notas
notas = [15, 18, 12, 19, 14, 16, 13, 17, 11, 20, 10, 9, 18, 15, 14]

# Validaciones
if len(notas) < 5 or len(notas) > 20:
    print("Error: Número de estudiantes debe estar entre 5 y 20")
else:
    mayor = max(notas)
    menor = min(notas)
    cantidad_mayor = notas.count(mayor)
    cantidad_menor = notas.count(menor)
    promedio = sum(notas) / len(notas)
    pares = sum(1 for n in notas if n % 2 == 0)
    impares = len(notas) - pares
    aprobados = sum(1 for n in notas if n >= 13)
    desaprobados = len(notas) - aprobados

    print(f"Mayor nota: {mayor}")
    print(f"Menor nota: {menor}")
    print(f"Cantidad con mayor nota: {cantidad_mayor}")
    print(f"Cantidad con menor nota: {cantidad_menor}")
    print(f"Promedio: {promedio:.2f}")
    print(f"Notas pares: {pares}")
    print(f"Notas impares: {impares}")
    print(f"Aprobados: {aprobados}")
    print(f"Desaprobados: {desaprobados}")
    print(f"% Aprobados: {(aprobados/len(notas))*100:.1f}%")
    print(f"% Desaprobados: {(desaprobados/len(notas))*100:.1f}%")
```
