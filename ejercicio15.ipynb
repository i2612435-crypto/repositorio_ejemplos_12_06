```python
# IDL2 - Ejercicio 15: Tipo de triángulo
# Estructura condicional

print("=== EJERCICIO 15: TIPO DE TRIÁNGULO ===")
l1, l2, l3 = 5, 5, 5

if l1 == l2 == l3:
    perimetro = l1 + l2 + l3
    print(f"Triángulo EQUILÁTERO")
    print(f"Perímetro: {perimetro}")
elif l1 == l2 or l2 == l3 or l1 == l3:
    print("Triángulo ISÓSCELES")
    if l1 == l2:
        print(f"Lados iguales: {l1} y {l2}")
    elif l2 == l3:
        print(f"Lados iguales: {l2} y {l3}")
    else:
        print(f"Lados iguales: {l1} y {l3}")
else:
    s = (l1 + l2 + l3) / 2
    area = (s*(s-l1)*(s-l2)*(s-l3)) ** 0.5
    print(f"Triángulo ESCALENO")
    print(f"Área: {area:.2f}")
```
