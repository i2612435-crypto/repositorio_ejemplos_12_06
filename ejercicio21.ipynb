```python
# Ejercicio 21: Suma y promedio de pares e impares del 1 al N
N = 10
pares = []
impares = []

for i in range(1, N+1):
    if i % 2 == 0:
        pares.append(i)
    else:
        impares.append(i)

print(f"Pares: suma={sum(pares)}, promedio={sum(pares)/len(pares):.2f}")
print(f"Impares: suma={sum(impares)}, promedio={sum(impares)/len(impares):.2f}")
```
