```python
# IDL2 - Ejercicio 16: Hora después de N segundos
# Estructura condicional

print("=== EJERCICIO 16: HORA + N SEGUNDOS ===")
hh, mm, ss = 14, 30, 45
n = 120

total_seg = hh*3600 + mm*60 + ss + n
nueva_hh = (total_seg // 3600) % 24
nueva_mm = (total_seg % 3600) // 60
nueva_ss = total_seg % 60

print(f"Hora original: {hh:02d}:{mm:02d}:{ss:02d}")
print(f"Agregar: {n} segundos")
print(f"Nueva hora: {nueva_hh:02d}:{nueva_mm:02d}:{nueva_ss:02d}")
```
